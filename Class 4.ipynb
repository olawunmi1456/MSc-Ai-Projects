```python
# Implement a Python function that accepts two lists and returns a third list containing the element-wise sums. 

# create function for element_wise_sum
def element_wise_sum(list1, list2):
    # If statement 
    if len(list1) != len(list2):
        return None
    return [a + b for a, b in zip(list1, list2)]

result1 = element_wise_sum([1, 2, 3], [4, 5, 6])
print(result1) 

result2 = element_wise_sum([1], [2, 3])
print(result2) 
```

```python
def element_wise_sum(**results):
    list1 = results.get('Arg1')
    list2 = results.get('Arg2')
    
    if list1 is None or list2 is None or len(list1) != len(list2):
        return None
    
    list_of_sums = [a + b for a, b in zip(list1, list2)]
    
    return (len(list_of_sums), list_of_sums)

# print result
result1 = element_wise_sum(Arg1=[1, 2, 3], Arg2=[4, 5, 6])
print(result1)

result2 = element_wise_sum(Arg1=[1], Arg2=[2, 3])
print(result2)
```

```python
# Implement a function that computes the Fibonacci sequence (recall that the sequence is constructed as Fi = Fi-2 + Fi-1). It will accept an argument 

#inititae fibonacci function
def fibonacci(n):
    if n <= 0:
        return [] 
    elif n == 1:
        return [0]  
    elif n == 2:
        return [0, 1]  
    
    fib_sequence = [0, 1]  
    
    for i in range(2, n):
        
        fib_sequence.append(fib_sequence[-1] + fib_sequence[-2])
    
    return fib_sequence
# print result
result = fibonacci(6)
print(result) 
```
