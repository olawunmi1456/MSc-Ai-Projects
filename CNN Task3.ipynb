```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# Define the input layer (shape matches MNIST image size: 28x28 pixels, 1 channel)
inputs = keras.Input(shape=(28, 28, 1))

# First convolutional layer (16 filters, 3x3 kernel, ReLU activation)
x = layers.Conv2D(filters=16, kernel_size=(3,3), activation='relu')(inputs)
# MaxPooling layer (pool size of 2)
x = layers.MaxPooling2D(pool_size=(2,2))(x)

# Second convolutional layer (64 filters, 3x3 kernel, ReLU activation)
x = layers.Conv2D(filters=64, kernel_size=(3,3), activation='relu')(x)
# Another MaxPooling layer (pool size of 2)
x = layers.MaxPooling2D(pool_size=(2,2))(x)

# Third convolutional layer (128 filters, 3x3 kernel, ReLU activation)
x = layers.Conv2D(filters=128, kernel_size=(3,3), activation='relu')(x)
# Another MaxPooling layer (pool size of 2)
x = layers.MaxPooling2D(pool_size=(2,2))(x)

# Flatten the feature maps
x = layers.Flatten()(x)

# Define the output layer (10 classes for digits 0-9, softmax activation)
outputs = layers.Dense(10, activation='softmax')(x)

# Define the model using the Functional API
model = keras.Model(inputs=inputs, outputs=outputs)

# Compile the model
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

# Load MNIST dataset
mnist = keras.datasets.mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Preprocess the data (normalize and reshape)
x_train = x_train.reshape(-1, 28, 28, 1).astype("float32") / 255.0
x_test = x_test.reshape(-1, 28, 28, 1).astype("float32") / 255.0

# Train the model
model.fit(x_train, y_train, epochs=5, batch_size=32, validation_data=(x_test, y_test))

# Evaluate the model
test_loss, test_acc = model.evaluate(x_test, y_test)
print("Test accuracy:", test_acc)

# Make predictions
predictions = model.predict(x_test)

# Display an example prediction
import numpy as np
example_index = 0
predicted_class = np.argmax(predictions[example_index])
print(f"Predicted class for example {example_index}: {predicted_class}")
```
