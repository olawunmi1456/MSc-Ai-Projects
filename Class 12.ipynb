```python
# Step 2a: Create 32 values of x between 0 and 1.571 (π/2)
x = np.linspace(0, 1.571, 32)

# Normalize x to be between 0 and 1
x_normalized = (x - np.min(x)) / (np.max(x) - np.min(x))

# Step 2b: Create y = cos(x)
y = np.cos(x)

# Reshape x to be a 2D array (required for Keras input)
x_normalized = x_normalized.reshape(-1, 1)

# Create the Keras ANN model
model = Sequential([
    Dense(5, activation='relu', input_shape=(1,)),  # Input layer + 1st hidden layer (5 neurons)
    Dense(3, activation='relu'),                    # 2nd hidden layer (3 neurons)
    Dense(1, activation='linear')                   # Output layer (1 neuron)
])

# Compile the model with Mean Squared Error loss and Adam optimizer
model.compile(optimizer=Adam(learning_rate=0.01), loss='mse', metrics=['mae'])

# Display the model summary
model.summary()

# Train the model for 200 epochs
history = model.fit(x_normalized, y, epochs=200, verbose=1)

# Plot the training loss over epochs
plt.plot(history.history['loss'])
plt.title('Training Loss Over Epochs')
plt.xlabel('Epochs')
plt.ylabel('Loss (MSE)')
plt.grid()
plt.show()
```

```python
# Step 2a: Create 32 values of x between 0 and 1.571 (π/2)
x = np.linspace(0, 1.571, 32)

# Normalize x to be between 0 and 1
x_normalized = (x - np.min(x)) / (np.max(x) - np.min(x))

# Step 2b: Create y = cos(x)
y = np.cos(x)

# Reshape x to be a 2D array (required for Keras input)
x_normalized = x_normalized.reshape(-1, 1)

# Create the Keras ANN model
model = Sequential([
    Dense(5, activation='relu', input_shape=(1,)),  # Input layer + 1st hidden layer (5 neurons)
    Dense(3, activation='relu'),                    # 2nd hidden layer (3 neurons)
    Dense(1, activation='linear')                   # Output layer (1 neuron)
])

# Compile the model with Mean Squared Error loss and Adam optimizer
model.compile(optimizer=Adam(learning_rate=0.01), loss='mse', metrics=['mae'])

# Display the model summary
model.summary()

# Train the model for 3000 epochs
history = model.fit(x_normalized, y, epochs=3000, verbose=1)

# Plot the training loss over epochs
plt.plot(history.history['loss'])
plt.title('Training Loss Over Epochs')
plt.xlabel('Epochs')
plt.ylabel('Loss (MSE)')
plt.grid()
plt.show()
```

```python
# Predict the values using the trained model
y_pred = model.predict(x_normalized)

# Plot the original training data and the model's predictions
plt.figure(figsize=(10, 6))
plt.scatter(x_normalized, y, color='blue', label='Original Data (y = cos(x))')
plt.plot(x_normalized, y_pred, color='red', linestyle='--', label='Model Predictions')
plt.title('Comparison of Training Data and Model Predictions')
plt.xlabel('Normalized x')
plt.ylabel('cos(x)')
plt.legend()
plt.grid()
plt.show()

# Calculate and display performance metrics
from sklearn.metrics import mean_squared_error, mean_absolute_error

mse = mean_squared_error(y, y_pred)
mae = mean_absolute_error(y, y_pred)

print(f"Mean Squared Error (MSE): {mse:.6f}")
print(f"Mean Absolute Error (MAE): {mae:.6f}")
```
