```python
# . Write a Python code that, given a list of names, creates a dictionary where each name 
#is grouped by the first letter. Your code should work for any list of names

names = ["Jim", "Hetty", "Kirsten", "Theo", "Henry", "Paul"]
grouped = {}

# Loop through each name in the list
for name in names:
    # Get the first letter of the name
    first_letter = name[0]
 # If this first letter is not already in the dictionary, create a new key for it
    if first_letter not in grouped:
        grouped[first_letter] = []  # Initialize an empty list for this key
    
    # Add the name to the list associated with its first letter
    grouped[first_letter].append(name)

# Display the dictionary with names grouped by their first letter
print(grouped)
```

```python
# 2. Given a dictionary, write a Python code that create a new dictionary where the keys are 
#used as values and the values are used as keys – e.g.: "computer" à "apple" becomes 
#"apple" à "computer" in the new dictionary. Your solution must work with any 
#dictionary

# Original dictionary
my_dict = {
    "computer": "apple",
    "model": "macbook pro",
    "processor": "M1 Pro",
    "year": 2021
}

# Dictionary to store the inverted key-value pairs
inverted_dict = {}

# Loop through each key-value pair in the original dictionary
for key, value in my_dict.items():
    # Swap the key and value, adding them to the new dictionary
    inverted_dict[value] = key

# Display the inverted dictionary
print(inverted_dict)
```

```python
# * A simple way to implement a Dictionary like type is to use a list of lists: the first level 
#determines the position of the element based on the “key”; the second level stores the 
#“value”. In order to determine the index of a particular “key” value, we can use the 
#hash() function and apply the modulo operator to limit the value within the range of the 
#list
# key = "alan"
# assuming that our inner list has 10 positions
#index = hash(key) % 10
#Since there is a limited number of elements on the list, some of the “key” values will 
#have a similar index. This is why each element of the list is another list: after 
#determining the index where the value should be added, the value is added to the list 
#at that index. The diagram below illustrates the process of adding the pair {"alan" : "7943" }

# Function to create a dictionary-like structure using a list of lists
def create_custom_dict(pairs, size=10):
    # Initialize the outer list with empty lists for each "bucket"
    custom_dict = [[] for _ in range(size)]
    
    # Loop through each key-value pair
    for key, value in pairs:
        # Determine the index using the hash of the key and the modulo operator
        index = hash(key) % size
        
        # Add the key-value pair to the appropriate list (bucket)
        custom_dict[index].append((key, value))
    
    return custom_dict

# Example key-value pairs to add to the custom dictionary
key_value_pairs = [("alan", "7943"), ("bob", "1234"), ("alice", "5678"), ("john", "91011")]

# Create the custom dictionary
custom_dict = create_custom_dict(key_value_pairs)

# Display the custom dictionary structure
print(custom_dict)
```
