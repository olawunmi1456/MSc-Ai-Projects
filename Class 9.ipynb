```python
import pandas as pd

# Load the dataset
df = pd.read_csv( "http://raptor.kent.ac.uk/~ds756/data/heart.csv")
df.head(10)
```

```python
#1. Using the dataframe idioma/cally, compute the sum of each feature. 
#Note, this only makes sense for some features. Skip over the ones for 
#which sum is not defined; use the Pandas mechanism for exclusion. The 

feature_sums = df.select_dtypes(include='number').sum()
print(feature_sums)
```

```python
#2. Compute the sum of the sums of the features.
sum_of_sums = feature_sums.sum()
print(sum_of_sums)
```

```python
#3. The mode is a measure of centre of a set (as are mean and median). It 
#is the element with the highest frequency. 
#mode ([4, 4, 1, 5, 5, 5, 7]) = 5  There are more instances of 5 
#Implement a func/on that computes the mode of a column. Name it 
#ComputeMode. It will be invoked as ComputeMode 
#(your_data_frame[“column name”]). You will have to do something to 
#the column before you call your func/on to ensure that it runs in linear /me. 
#Demonstrate correctness by using it to compute the mode of “Age”. 

def ComputeMode(column):
   #compute mode in linear
    return column.mode().iloc[0]

# the 'Age' column 
age_mode = ComputeMode(df['Age'])
print(age_mode)
```

```python
#4. Compute the rate of heart disease by age. 
heart_disease_rate_by_age = df.groupby('Age')['HeartDisease'].mean()
print(heart_disease_rate_by_age)
```

```python
#5.Create a new column, called “AgeNorm,” that is the age column divided 
#by the max age.
AgeNorm = df['Age'] / df['Age'].max()
print(AgeNorm)
```
