```python
import pandas as pd
from statistics import mean as compute_mean

# Load the dataset
df = pd.read_csv("http://raptor.kent.ac.uk/~ds756/data/heart.csv")

# Display the first few rows 
print(df.head())
```

```python
# 1. How many rows does it have?
num_rows = df.shape[0]
print(f"Number of rows: {num_rows}")
```

```python
# 2. How many features does it have?
num_features = df.shape[1]
print(f"Number of features: {num_features}")
```

```python
# 3. How many features are of type 'int64'?
int64_features = df.select_dtypes(include=["int64"]).columns
num_int64_features = len(int64_features)
print(f"Number of features of type 'int64': {num_int64_features}")
```

```python
# 4. What is the mean Oldpeak?
mean_oldpeak = compute_mean(df["Oldpeak"])
print(f"Mean Oldpeak: {mean_oldpeak}")
```

```python
# 5. What is the mean RestingBP?
mean_restingbp = compute_mean(df["RestingBP"])
print(f"Mean RestingBP: {mean_restingbp}")
```

```python
# 6. What is the mean Age?
mean_age = compute_mean(df["Age"])
print(f"Mean Age: {mean_age}")
```

```python
# 7. What percentage of the cases developed heart disease?
heart_disease_percentage = (df["HeartDisease"].sum() / num_rows) * 100
print(f"Percentage of cases with heart disease: {heart_disease_percentage:.2f}%")
```

```python
# 8. Index the Age feature
age_feature = df["Age"]
print(f"Indexed Age feature:\n{age_feature}")
```

```python
# 9. Create a subset dataframe for people in their 40's
forties_subset = df[(df["Age"] >= 40) & (df["Age"] < 50)]

# a. Mean Oldpeak for people in their 40's
mean_oldpeak_forties = compute_mean(forties_subset["Oldpeak"])
print(f"Mean Oldpeak (40's): {mean_oldpeak_forties}")

# b. Mean RestingBP for people in their 40's
mean_restingbp_forties = compute_mean(forties_subset["RestingBP"])
print(f"Mean RestingBP (40's): {mean_restingbp_forties}")

# c. Mean Age for people in their 40's
mean_age_forties = compute_mean(forties_subset["Age"])
print(f"Mean Age (40's): {mean_age_forties}")
```
